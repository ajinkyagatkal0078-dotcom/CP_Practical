```python
# If elif else statement for marks of the student
# >= 95: Grade A, 90-94: Grade B, 85-89: Grade C, 75-84: Grade D, 65-74: Grade E, below 65: Grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("Grade A")
elif marks >= 90:
    print("Grade B")
elif marks >= 85:
    print("Grade C")
elif marks >= 75:
    print("Grade D")
elif marks >= 65:
    print("Grade E")
else:
    print("Grade F")
```
